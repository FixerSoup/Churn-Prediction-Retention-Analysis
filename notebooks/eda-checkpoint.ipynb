```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
location =  Path(r"..\cleaned.csv")
df=  pd.read_csv(location)
```

```python
df.head()
```

### Churn per age

```python
age_churn = df.groupby('age')['churn_risk_score'].mean()
age_churn 
```

```python
plt.figure(figsize = (14,6));
plt.plot(age_churn.index,age_churn.values,marker = 'o',color = 'green');
plt.xlabel('age');
plt.ylabel('Churn rate');
plt.title('Churn rate by age');
plt.grid(True, alpha=0.3)
```

From the above line plot we can infer that maximum number of customers who churn are at the age group of between 36-38 years of age!

### Churn per gender

```python
gender_churn = df.groupby('gender')['churn_risk_score'].mean()
gender_churn
```

```python
plt.figure(figsize = (14,6));
plt.barh(gender_churn.index,gender_churn.values,color = 'blue');
plt.xlabel('gender');
plt.ylabel('Churn rate');
plt.title('Churn rate by gender');
plt.grid(True, alpha=0.3)
```

From the above bar plot we can infer that most of the gender detail is not available for churn prediction

### Churn per region_category

```python
region_churn = df.groupby('region_category')['churn_risk_score'].mean()
region_churn
```

```python
plt.figure(figsize = (14,6));
plt.barh(region_churn.index,region_churn.values,color = 'orange');
plt.xlabel('Region');
plt.ylabel('Churn rate');
plt.title('Churn rate by Region');
plt.grid(True, alpha=0.3)
```

From the available data city area customers contain the most churns

### Churn per membership_category

```python
member_churn = df.groupby('membership_category')['churn_risk_score'].mean()
member_churn
```

```python
plt.figure(figsize = (14,6));
plt.barh(member_churn.index,member_churn.values,color = 'magenta');
plt.xlabel('Membership Category');
plt.ylabel('Churn rate');
plt.title('Churn rate by Membership');
plt.grid(True, alpha=0.3)
```

Premium and Platinum member have 0 Churn rate!

```python
print(df['joining_date'].dtype)
```

```python
df['joining_date'] = pd.to_datetime(df['joining_date'], format='%Y-%m-%d', errors='coerce')
```

```python
df['joining_date']
```

### churn by joining date

```python
df['joining_month'] = df['joining_date'].dt.to_period('M')
trend = df.groupby('joining_month')['churn_risk_score'].mean()

plt.figure(figsize=(12, 5))
plt.plot(trend.index.astype(str), trend.values, marker='o')
plt.xlabel('Joining month')
plt.ylabel('Average churn rate')
plt.title('Churn rate by joining cohort')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

from this we can infer that most of the churn happened between june 2016 and july 2016

### Chun per joined_through_referral

```python
referral_churn = df.groupby('joined_through_referral')['churn_risk_score'].mean().reset_index()
referral_churn
```
