```python
def find_counts(df : pd.DataFrame):
    for col,data in names.items():
        if data == str :
            print('------------------------------------------')
            print(df[col].value_counts().reset_index(),'\n\n')
            print('------------------------------------------')
        else:
            print('------------------------------------------')
            print(df[col].value_counts().reset_index(),'\n\n')
            print('------------------------------------------')
```

```python
def smart_fillna(df: pd.DataFrame):
    for col,data in names.items():
        if data == str:
            print('-----------------------------------------')
            df[col] = df[col].fillna('No data available')
            print(f'Filled {col} with No data available')
        else:
            print('-----------------------------------------')
            fill = df[col].median()
            df[col] = df[col].fillna(fill)
            print(f'Filled {col} with {fill}')
```

```python
smart_fillna(df)
```
